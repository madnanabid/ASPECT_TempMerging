```python
def compute_probabilities(df, member_cols, threshold):
    members_binary = (df[member_cols] > threshold).astype(int)
#    print(members_binary.mean(axis=1).values, threshold)
#    print("---",  (df["obs_FFD"] > threshold).astype(int).values)
    return members_binary.mean(axis=1).values, (df["obs_FFD"] > threshold).astype(int).values
```

```python
def relative_economic_value_curve(forecast_probs, observations, n_points=1000):
    forecast_probs = np.asarray(forecast_probs)
    observations = np.asarray(observations)
    print("era5=",observations)

    loss = 1.0
    climatology = observations.mean()
    print(climatology)
    cost_loss_ratios = np.linspace(0.01, 0.99, n_points)
    values = []

    for r in cost_loss_ratios:
       # print('check=',r)
        cost = r * loss

        # Forecast 
        action_fcst = forecast_probs > r
        expense_fcst = np.mean(np.where(action_fcst, cost, np.where(observations == 1, loss, 0)) )

        # Climatology
        if climatology > r:
            expense_clim = cost
          #  print(cost, expense_clim)
        else:
            expense_clim = np.mean( np.where(observations == 1, loss, 0) )

        # Perfect forecast expense 
        expense_perf = np.mean( np.where(observations == 1, cost, 0) )
       # print('expense_perf=', expense_perf)
        # ---- Relative economic value ----
        if expense_clim > expense_perf:
            V = (expense_clim - expense_fcst) / (expense_clim - expense_perf)
            V = max(V, 0.0)
        else:
            V = np.nan
        values.append(V)

    return cost_loss_ratios, np.array(values)
```

```python
def bootstrap_rev(forecast_probs,observations,n_bootstrap=2000,n_points=100,ci=(2.5, 97.5), random_state=40):
    rng = np.random.default_rng(random_state)
    forecast_probs = np.asarray(forecast_probs)
    observations = np.asarray(observations)

    n = len(observations)
    rev_boot = np.zeros((n_bootstrap, n_points))

    for i in range(n_bootstrap):
        idx = rng.integers(0, n, n)  # paired resampling
        fp_b = forecast_probs[idx]
        obs_b = observations[idx]

        _, rev_b = relative_economic_value_curve(fp_b, obs_b, n_points=n_points)
        rev_boot[i, :] = rev_b
    r, rev_mean = relative_economic_value_curve(forecast_probs, observations, n_points=n_points)

    rev_lower = np.nanpercentile(rev_boot, ci[0], axis=0)
    rev_upper = np.nanpercentile(rev_boot, ci[1], axis=0)

    return r, rev_mean, rev_lower, rev_upper
```
